### 03.Handling missing values

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
df.astype(str).apply(lambda X:X.str.strip()=='').sum()
```

```python
df = df[~df['TotalCharges'].astype(str).str.strip().eq('')]
```

```python
(df.isnull() | df.astype(str).apply(lambda X:X.str.strip()=='')).sum()
```

### 04.Handling duplicate values

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

### 05.Drop columns

```python
df = df.drop(columns=['customerID'])
df.head(5)
```
