### 01.Import dependencies

```python
import pandas as pd
```

### 02.Import dataset

```python
df = pd.read_csv("processed/handling_duplicate_and_missing_values.csv")
df.head()
```

### 03. Feature Binning

```python
def tenure_binning(tenure):
    if tenure <= 12:
        return 'Newer'
    elif tenure <=48:
        return 'Medium'
    else:
        return 'Long-term'
```

```python
df['tenure_bins'] = df['tenure'].apply(tenure_binning)
df.head(5)
```

```python
del df['tenure']
```
